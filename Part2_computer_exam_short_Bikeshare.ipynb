## The data

The `Bikeshare` data set of the `ISLP` package contains the hourly number of users of a bike sharing program in Washington, D.C., in 2011: each of the 8,645 observations is one hour of one day. **The goal is to predict the number of bikers (`bikers`) during a given hour as accurately as possible.**

| Variable | Description |
|---|---|
| `season` | Season of the year, coded 1 (winter), 2 (spring), 3 (summer), 4 (fall) |
| `mnth` | Month of the year (12 categories) |
| `day` | Day of the year, from 1 to 365 |
| `hr` | Hour of the day, from 0 to 23 (24 categories) |
| `holiday` | 1 if the day is a holiday, 0 otherwise |
| `weekday` | Day of the week, coded from 0 (Sunday) to 6 (Saturday) |
| `workingday` | 1 if the day is neither a weekend nor a holiday, 0 otherwise |
| `weathersit` | Weather situation (4 categories: `clear`, `cloudy/misty`, `light rain/snow`, `heavy rain/snow`) |
| `temp` | Normalized temperature (between 0 and 1) |
| `atemp` | Normalized feeling temperature (between 0 and 1) |
| `hum` | Normalized humidity (between 0 and 1) |
| `windspeed` | Normalized wind speed (between 0 and 1) |
| `casual` | Number of casual (non-registered) bikers during the hour |
| `registered` | Number of registered bikers during the hour |
| `bikers` | Total number of bikers during the hour — **the response** |
